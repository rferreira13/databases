```python
import os
import json
from pathlib import Path
from getpass import getpass

# 2.1) Captura "secreta" (entrada não é exibida)
username = getpass("Kaggle username: ").strip()
api_key  = getpass("Kaggle API key: ").strip()

# 2.2) Diretório ~/.kaggle no HOME do usuário atual (portável, inclusive em Databricks)
kaggle_dir = Path.home() / ".kaggle"
kaggle_dir.mkdir(parents=True, exist_ok=True)

# 2.3) Caminho do arquivo de credenciais
kaggle_json = kaggle_dir / "kaggle.json"

# 2.4) Escreve as credenciais no JSON
creds = {"username": username, "key": api_key}
with open(kaggle_json, "w") as f:
    json.dump(creds, f)

# 2.5) Permissões: 600 em sistemas tipo Unix (Linux/macOS/WSL/Databricks)
try:
    os.chmod(kaggle_json, 0o600)
except Exception as e:
    # Em Windows puro, chmod pode não aplicar do mesmo modo — este aviso é esperado
    print(f"Aviso: não foi possível aplicar chmod 600 ({e}). Em Windows isso é esperado.")

print(f"Credenciais gravadas com segurança em: {kaggle_json}")
```

```python
from kaggle.api.kaggle_api_extended import KaggleApi

api = KaggleApi()
api.authenticate()  # usa ~/.kaggle/kaggle.json ou env vars

# Exemplo: baixar e descompactar um dataset
api.dataset_download_files(
    dataset="zillow/zecon",  # owner/slug
    path="dados_kaggle",
    unzip=True
)
```
